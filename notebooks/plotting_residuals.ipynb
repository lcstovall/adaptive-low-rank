# Plotting residuals notebook

This notebook visualizes the normalized residual curves saved by the experiment runs. Each cell loads the saved results for one dataset and produces the corresponding figure in the `figures/` directory.


```python
from pathlib import Path
import importlib.util
import pickle
import yaml

repo_root = Path.cwd()
if not (repo_root / "results").exists():
    repo_root = repo_root.parent

plotting_path = repo_root / "src" / "adaptive_low_rank" / "plotting.py"
spec = importlib.util.spec_from_file_location("adaptive_low_rank_plotting", plotting_path)
plotting = importlib.util.module_from_spec(spec)
spec.loader.exec_module(plotting)


def load_results(results_dir):
    results_dir = Path(results_dir)
    with (results_dir / "results.pkl").open("rb") as file:
        results = pickle.load(file)
    with (results_dir / "config.yml").open() as file:
        config = yaml.safe_load(file)
    dataset_name = config.get("name", config.get("dataset", results_dir.name))
    return results, dataset_name
```

```python
results, experiment_name = load_results(repo_root / "results" / "interactions")
plotting.plot_residuals(
    results,
    output_dir=repo_root / "figures",
    name=experiment_name,
)
```
