```python
# Load the plotting module directly from the local source tree.
from pathlib import Path
import importlib.util
import pickle
import yaml

repo_root = Path.cwd()
if not (repo_root / "results").exists():
    repo_root = repo_root.parent

plotting_path = repo_root / "src" / "adaptive_low_rank" / "plotting.py"
spec = importlib.util.spec_from_file_location("workspace_plotting", plotting_path)
plotting = importlib.util.module_from_spec(spec)
spec.loader.exec_module(plotting)
```

```python
# Interactions
results_dir = repo_root / "results" / "interactions_residual"

with (results_dir / "results.pkl").open("rb") as file:
    results = pickle.load(file)

with (results_dir / "config.yml").open() as file:
    experiment_name = yaml.safe_load(file)["dataset"]

plotting.plot_residuals(
    results,
    output_dir=repo_root / "figures",
    name=experiment_name,
)
```
