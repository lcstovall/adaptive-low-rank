# Plotting alpha values notebook

This notebook plots the alpha trajectories for the alpha value experiments.


```python
from pathlib import Path
import pickle
import yaml

from adaptive_low_rank.plotting import plot_alphas

repo_root = Path.cwd()
if not (repo_root / "results").exists():
    repo_root = repo_root.parent


def load_results(results_dir):
    results_dir = Path(results_dir)
    with (results_dir / "results.pkl").open("rb") as file:
        results = pickle.load(file)
    with (results_dir / "config.yml").open() as file:
        config = yaml.safe_load(file)
        dataset_name = config["dataset"]
    return results, dataset_name
```

```python
results, experiment_name = load_results(repo_root / "results" / "interactions")
plot_alphas(results, repo_root / "figures", experiment_name)
```
