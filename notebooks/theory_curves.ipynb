# Theory curves notebook

This notebook visualizes the theory and empirical residual curves for selected datasets.

```python
from pathlib import Path
import importlib.util
import pickle

repo_root = Path.cwd()
if not (repo_root / "results").exists():
    repo_root = repo_root.parent

theory_path = repo_root / "scripts" / "generate_thoery_curves.py"
spec = importlib.util.spec_from_file_location(
    "generate_thoery_curves", theory_path
)
theory = importlib.util.module_from_spec(spec)
spec.loader.exec_module(theory)


def load_results(results_dir):
    results_dir = Path(results_dir)
    with (results_dir / "results.pkl").open("rb") as file:
        results = pickle.load(file)
    return results, theory.load_dataset(results_dir.name)
```

```python
results, X = load_results(repo_root / "results" / "interactions")
n_candidates = 10
max_k = 120
alphas = theory.compute_batchmax_alphas(results, n_candidates=n_candidates)

fig, ax = theory.plot_theory_curves(
    X,
    max_k=max_k,
    alphas=alphas,
    results=results,
    batchmax_n_candidates=n_candidates,
    output_path=repo_root / "figures" / "interactions_theory.png",
)
```
